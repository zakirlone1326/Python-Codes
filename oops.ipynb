```python
#constructor
class Student:
 
    def __init__(self, name): #self is refrence to current instance of class
        self.name = name
        print("adding new student in DB ")
s1 = Student("zakir")
print(s1.name)
s2 = Student("irfan")
print(s2.name)
```

```python
#constructor
class Student:

    def __init__(self): #default constructer it includes only self
        pass
       #parameterized constructor means parameters other than self
    def __init__(self, name, marks): #self is refrence to current instance of class
        self.name = name #self.parameter isused when parameters for every object is diff, since name of every student is diff thus we write self.name if it was same for all then we simply write name
        self.marks = marks
        print("adding new student in DB ")
s1 = Student("zakir",30)
print(s1.name, s1.marks)
s2 = Student("irfan",50)
print(s2.name, s2.marks)
```

```python
class Student:
    college = "North campus " #this is same for all that is why we write it here not in slef
    def __init__(self, name , marks):
        self.name= name
        self.marks = marks
s1 = Student("zakir",30)
print(s1.name, s1.marks, s1.college)

s2 = Student("irfan", 50)
print(s2.name, s2.marks, s2.college)
```
