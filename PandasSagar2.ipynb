```python
mca = {
    "Name": ["zakir", "irfan", "ishtiyaq", "salim", "inam"],
    "Roll No": [11, 6, 21, 7, 3 ],
    "District": ["kupwara", "Baramulla", "Bandipora", "Rafiabad", "Sopore"]
}
print(mca)
```

```python
import pandas as pd
df = pd.DataFrame(mca)
print(df)
```

```python
df["Roll No"] = df["Roll No"] + 2
print(df)
```

```python
df.insert(2, "Score", [70,60,None,55,66]) #inserting new column
print(df)
```

```python
df.loc[3, "Score"]= 58  #updateing score of index 3
print(df)
```

```python
df["Roll No"]= df["Roll No"] - 2 #updating roll no of all
print(df)
```

```python
#removing columns
df.drop(columns = ["District"], inplace = True)
print(df)
```

```python
#handling missing values
print(df.isnull()) #check if any data is missing
print(df.isnull().sum()) #tells how many valus are missing
```

```python
#updating this missing value now
df.dropna(axis = 0, inplace = True) #removes whole row in which data is missing
print(df) #axis = 0 is for rows and axis = 1 is for columns
```

```python
#sorting data
ZM = {
    "Name": ["kupwara", "sopore", "handwara", "pulwama", "bandipora", "pampore"],
    "Sale": [13, 13, 9, 10, 10, 9],
    "District": ["kupwara", "baramulla", "kupwara", "pulwama", "bandipora", "pulwama"]
}
import pandas  as pd
df = pd.DataFrame(ZM)
print(df)
```

```python
#grouping
grouped= df.groupby("District")["Sale"].sum() #("District means" throught what you want to group them and "Sale means "throught what you wan tto check)
print(grouped)
```

```python
#grouping more than one column at once
grouped = df.groupby(["District","Sale"])["Name"].sum()
print(grouped)
```

```python
grouped2= df.groupby("District")["Sale"].mean()
print(grouped2)
```

```python
#merging 
import pandas as pd
df_customers = pd.DataFrame({
    "customerID":[1,2,3,4],
    "Name": ["zakir", "irfan", "ishtiyaq", "musadiq"]
})


df_orders = pd.DataFrame({
    "customerID": [1,2,3,5],
    "order amount": [200,300,450,240]
})
df_merged = pd.merge(df_customers, df_orders, on="customerID", how = "inner")
print(df_merged)  #"inner" means only those will be print whose keys matches
```

```python
df_merged= pd.merge(df_customers,df_orders, on ="customerID", how="outer")
print(df_merged) #outer prints all
```

```python
df_merged = pd.merge(df_customers, df_orders, on= "customerID", how = "left")
print(df_merged)
```

```python
#concatanation of columns
import pandas as pd
df_city1 = pd.DataFrame({
    "customerID":[1,2,],
    "Name": ["zakir", "irfan"]
})


df_city2 = pd.DataFrame({
    "customerID": [3,5],
    "Name": ["ishtiyaq", "musadiq"]
})
df_concat = pd.concat([df_city1, df_city2], axis =1, ignore_index= True)
print(df_concat) #axis = o means verticlly amd =1 means horizentally
```
