```python
#armstrong number check
#armstrong num means sum of (each digit ^ number of digits) = the number itself

num = 370
sum = 0
temp = num

while temp >0:
    digit = temp % 10
    cube = digit**3
    sum += cube
    temp //=10

if sum == num:
    print("armstrong num")
else:
    print("Not armstrong")
```
